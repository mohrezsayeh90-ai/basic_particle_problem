# Particle PINN study
فایل ZIP پروژه را باز کنید و این نوت‌بوک را در پوشهٔ پروژه اجرا کنید. اجرای quick برای بررسی کامل مسیر اجراست؛ معیارهای دقت را در REPORT.md بخوانید.

```python
from pathlib import Path
import sys, subprocess
root = Path.cwd()
assert (root / "run_particle_pinn.py").exists(), "Extract the project ZIP and change to its directory first."
subprocess.check_call([sys.executable, "-m", "pip", "install", "-r", "requirements.txt"])
```

```python
from datetime import datetime, timezone
out = root / "results" / datetime.now(timezone.utc).strftime("notebook_%Y%m%dT%H%M%S_%fZ")
command = [sys.executable, "run_particle_pinn.py", "--preset", "quick", "--out", str(out)]
result = subprocess.run(command, cwd=root, check=False)
print("Exit:", result.returncode, "Results:", out)
assert (out / "metrics.json").exists(), "Read the stage failure above."
```

```python
import json
from IPython.display import display, Image, Markdown
metrics = json.loads((out / "metrics.json").read_text())
print(json.dumps(metrics["quality_gate"], indent=2))
display(Markdown((out / "REPORT.md").read_text()))
for name in ["02_architecture.png", "04_flow_fields.png", "10_particle_paths.png", "19_3d_benchmark.png"]:
    file = out / "figures" / name
    if file.exists(): display(Image(filename=str(file), width=1000))
gif = out / "figures" / "17_particle_motion.gif"
if gif.exists(): display(Image(filename=str(gif)))
```
